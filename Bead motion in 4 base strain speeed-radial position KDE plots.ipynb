```python
import numpy as np
import tifffile
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
file_path = r"C:\Users\ayantis\Desktop\PhD\Research\Project 1 - Formation of a swelling gel underlies a morphological transition in Bacillus subtilis biofilms\PNAS submission\Revision\Review experiments\Review expt #7\WT_tracks.xlsx"
df_WT = pd.read_excel(file_path, header=1)
WT_mean_speed_10 =  pd.to_numeric(df_WT["Track mean speed"].iloc[4:],errors='coerce')*(30/116)
WT_mean_X_10 =  pd.to_numeric(df_WT["Track mean X"].iloc[4:],errors='coerce')
```

```python
figsize_mm = 80
figsize_in = figsize_mm / 25.4  # = 1.496 inches

fig, ax = plt.subplots(figsize=(figsize_in, figsize_in))
sns.kdeplot(x = WT_mean_X_10, y= WT_mean_speed_10, fill=True, thresh=0.05, levels=4,
        bw_adjust=0.7, alpha=0.9, cmap='Blues',
        linewidths=0.4, ax=ax);
ax.set_ylabel(r"Mean tracked speed ($\mu$m/s)")
ax.set_xlabel("Mean radial distance (µm)")
sns.despine()
ax.set_title('WT')
plt.savefig(r"C:\Users\ayantis\Desktop\PhD\Research\Project 1 - Formation of a swelling gel underlies a morphological transition in Bacillus subtilis biofilms\PNAS submission\Revision\Review experiments\Review expt #7\WT KDE plot.svg", format = "svg", dpi =600)
plt.show()
```
