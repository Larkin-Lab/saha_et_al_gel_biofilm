```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import numpy.ma as ma
```

```python
q = np.zeros(25)
for i in range(1,26):
    address_original = r'C:\Users\ayantis\Desktop\PhD\Research\Experimental Data\F0825\F0825 Stereoscope data\N=1\B'+str(i)+" cropped.tif"
    image_original = cv2.imread(address_original, cv2.IMREAD_GRAYSCALE)
    
    image_binary = ((image_original > 0 ).astype(np.uint8))
    
    address_wrinkles = r'C:\Users\ayantis\Desktop\PhD\Research\Experimental Data\F0825\F0825 Stereoscope data\N=1\B'+str(i)+" wrinkles.tif"
    image_wrinkles = cv2.imread(address_wrinkles, cv2.IMREAD_GRAYSCALE)
   
    q[i-1] = (np.sum(image_wrinkles)/255)/(np.sum(image_binary))
    
    fig, ax = plt.subplots(nrows = 1, ncols = 3, figsize = (15, 5))
    ax[0].imshow(image_original, cmap = 'gray')
    ax[0].set_title("Original Biofilm "+str(i))
    
    ax[1].imshow(image_binary, cmap = 'gray')
    ax[1].set_title("Binary Biofilm mask")
    
    ax[2].imshow(image_wrinkles, cmap = 'gray')
    ax[2].set_title(f"Thresholded wrinkles with q = {q[i-1]}")
    
    for a in ax:
        a.axis('off')
        
    plt.tight_layout()
    plt.show()
```
